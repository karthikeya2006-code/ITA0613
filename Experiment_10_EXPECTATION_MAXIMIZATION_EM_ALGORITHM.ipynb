# EXPERIMENT NUMBER 10: EXPECTATION-MAXIMIZATION (EM) ALGORITHM

**Aim:** To implement the Expectation-Maximization algorithm for clustering.

```python
from sklearn.datasets import make_blobs
from sklearn.mixture import GaussianMixture
import matplotlib.pyplot as plt

# Generate Dataset
X, _ = make_blobs(
    n_samples=300,
    centers=3,
    cluster_std=0.60,
    random_state=42
)

# EM Algorithm
gmm = GaussianMixture(
    n_components=3,
    random_state=42
)

gmm.fit(X)

labels = gmm.predict(X)

# Plot Clusters
plt.scatter(
    X[:,0],
    X[:,1],
    c=labels,
    cmap="viridis"
)

plt.title("Expectation Maximization Clustering")
plt.show()
```
