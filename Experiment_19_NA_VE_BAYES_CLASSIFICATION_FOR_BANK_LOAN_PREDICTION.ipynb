# EXPERIMENT NUMBER 19: NAÏVE BAYES CLASSIFICATION FOR BANK LOAN PREDICTION

**Aim:** To implement Naïve Bayes classification for Bank Loan prediction.

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score

# Sample Dataset
data = {
    "Income":[30,40,50,60,70,80,90],
    "CreditScore":[650,700,720,680,750,800,820],
    "LoanApproved":[0,0,1,1,1,1,1]
}

df = pd.DataFrame(data)

X = df[["Income","CreditScore"]]
y = df["LoanApproved"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

model = GaussianNB()
model.fit(X_train, y_train)

prediction = model.predict(X_test)

print("Accuracy:", accuracy_score(y_test, prediction))

# New Prediction
sample = [[55,710]]
print("Loan Approval:", model.predict(sample)[0])
```
